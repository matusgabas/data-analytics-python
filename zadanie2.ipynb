```python
import pandas as pd
import matplotlib.pyplot as plt

t = pd.read_csv("../titanic.csv", sep=";")
t.head(10)
```

```python
t.shape
```

```python
print(t['cestovne'].mean().round(2))
print(t['cestovne'].median())
print(t['cestovne'].mode()[0])
```

```python
t['cestovne'].plot(kind='hist', bins=50, edgecolor='white')
plt.axvline(t['cestovne'].mean(), color='red', label='priemer')
plt.axvline(t['cestovne'].median(), color='green', label='median')
plt.axvline(t['cestovne'].mode()[0], color='yellow', label='modus')
plt.xlabel('cestovne')
plt.ylabel('pocet cestujucich')
plt.legend()
plt.show()
```
